##### Rules for defining the Variable

(a.)    A Variable name always contains letters, numbers and underscore.

```python
first_name1 = 'Parag'
print(first_name1)
```

(b.)    It should always start with letter or an underscore

```python
name = 'Parag'   ##### Variable beginning with letter
print(name)
_name = 'Parag'  ##### Variable beginning with underscore
print(_name)
```

(c.) It should contain underscore if needed

```python
first_name = 'Parag'  ##### Underscore is present in between the variable names
print(first_name)
```

(d.)    It should not have hyphen in between

```python
#first-name = 'Parag'
#print(first-name)
```

(e.)    It should not be the name of the keyword

```python
#def = 10
#print(def)
```

(f.)    It should not start with digit

```python
#1name = 'Parag'
#print(1name)
```

(g.)    It should not have any special symbols like ($,%,&)etc

```python
#$name = 'Parag'   ##### Variable should not have '$'
#%name = 'Parag'   ##### Variable should not have '%'
#&name = 'Parag'   ##### Variable should not have '&'
```

##### Properties of Variable

(a.)    A Variable is case sensitive. It treats the upper and lower case differently.

```python
a = 10
print(a)

A = 15
print(A)
```

##### OBSERVATIONS:

1.  a -------> 10, A --------> 15. This is because Python treats a and A both differently and both have the different meanings.

So a and A product different results.

(b.)    A Variable follows snake convention.

```python
first_name = 'Parag'
last_name  = 'Pujari'

print(first_name)
print(last_name)
```

40. You have the following variables:

original_price = 1000

discount = 200

tax = 80

Calculate the final price using only variables and arithmetic.

The logic is:

Price after discount = original price - discount

Final price = price after discount + tax

```python
original_price = 1000

discount = 200

tax = 80

price_after_discount = original_price - discount
print(f'Price after discount is {price_after_discount}')

final_price = price_after_discount + tax
print(f'Final price is {final_price}')
```
