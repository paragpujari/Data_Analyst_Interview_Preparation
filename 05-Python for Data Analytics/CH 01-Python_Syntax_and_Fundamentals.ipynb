11. Write a Python program to accept a number and check whether it is a palindrome number

```python
num = int(input("Enter the number"))

x = num

rev=0

while(x!=0):
    d=x%10
    rev=(rev*10)+d
    x=x//10
print(f'Reverse of the number is {rev}')

if(rev==num):
    print(f"{num} is palindrome")
else:
    print(f"{num} is not palindrome")
```

12.  Write a Python program to accept an integer and determine whether it is an Armstrong number

```python
num = int(input("Enter the number"))

x = num
s=0
while(x!=0):
   d = x%10
   s = s + (d*d*d)
   x = x // 10
print(f'Sum of all the digits of the number is {s}')

if(s==num):
    print(f"{num} is armstrong")
else:
    print(f"{num} is not armstrong")
```

16. Employee Salary Calculation: An organization calculates employee salary based on basic salary. Write a Python program to calculate HRA, DA, gross salary and net salary using variables, arithmetic operators and conditional statements. Assume suitable percentage values

Formulas:

Calculation	Formula

HRA:    	        Basic Salary × 20 / 100

DA:     	        Basic Salary × 10 / 100

Gross Salary:   	Basic Salary + HRA + DA

PF:             	Basic Salary × 12 / 100

Professional Tax:    ₹200

Net Salary:     	Gross Salary − PF − Professional Tax

```python
##### enter the basic salary
basic_salary = int(input("Enter the basic salary"))

##### Calculate the HRA
hra          =  basic_salary*(20/100)
da           =  basic_salary*(10/100)
gross_salary =  basic_salary+hra+da

pf           =  basic_salary*(12/100)

tax          =  200

net_salary   = gross_salary - (pf + tax)

print(f"New salary is {net_salary}")
```

17.  Bank Transaction Validation: A bank allows withdrawals only when the withdrawal amount is positive and does not exceed the available account balance. Write a Python program to accept the balance and withdrawal amount and display the transaction status

```python
##### enter the balance
balance = int(input("Enter the balance"))

withdrawl_amount = int(input("Enter the withdrawl amount"))

if((withdrawl_amount>0)and(balance>withdrawl_amount)):
    print("Valid Transaction")
    remaining_balance = balance-withdrawl_amount
    print(f"Remaining amount balance is {remaining_balance}")
elif(withdrawl_amount<0):
    print("Invalid Transaction")
elif(withdrawl_amount>balance):
    print("Withdrawl amount exceeds the available balance")
else:
    print("No status")
   
```

20. Daily Sales Target Analysis: A company has a daily sales target of ₹10,000. Write a Python program to accept sales values for seven days, calculate total sales and average sales, and determine the number of days the target was achieved

```python
target = 10000
total_sales = 0
target_count = 0

for day in range(1, 8):
    sales = float(input(f"Enter sales for Day {day}: "))

    total_sales = total_sales + sales

    if sales >= target:
        target_count = target_count + 1

average_sales = total_sales / 7

print("\n----- Weekly Sales Report -----")
print("Total Sales:", total_sales)
print("Average Sales:", average_sales)
print("Target Achieved Days:", target_count)
print("Target Not Achieved Days:", 7 - target_count)
```
