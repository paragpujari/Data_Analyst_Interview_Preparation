##### Operators

Operators are the special signs andsymbols that is used to perform certain operations in the program.


##### Main Operators Categories

---------------------------------------------------------------------------
Category	|        Operators	          |         Purpose                   

Arithmetic	|   + - * / // % **	          |   Mathematical calculations

Comparison	|    == != > < >= <=	      |      Compare values

Assignment	|    = += -= *= /= //= %= **= |	Assign/update values

Logical	    |    and or not	              |  Combine conditions

Bitwise	    |    & \| ^ ~ << >>           |	Work at bit level

Membership	|    in, not in	              |  Check membership

Identity	|   is, is not	              |  Check object identity


---------------------------------------------------------------------------

##### Types of Python Operators

1.  Arithmetic Operators

```python
a = 10
b = 3

print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a//b)
print(a%b)
print(a**b)
```

2. Comparision Operators

```python
a = 10
b = 3

print(a==b)
print(a!=b)
print(a>b)
print(a<b)
print(a>=b)
print(a<=b)
```

3.  Assignment Operators

```python
##### performing the assignment operations
a = 10
a += 5
print(a)

b = 10
b -= 5
print(b)

c = 5
c *= 2
print(c)

d = 10
d /= 2
print(d)

e = 4
e //= 3
print(e)

f = 5
f %= 2
print(f)

g = 10
g **= 3
print(g)
```

4. Logical Operators

```python
print((10 > 5) and (8 > 6))

print((2 > 5) or (10 > 13))

print(not(10>5) or (9>8))
```

5. Bitwise Operators

```python
a = 5
b = 3

##### perform bitwise AND
print(a & b)

##### perform bitwise OR
print(a | b)

##### perform bitwise XOR
print(a ^ b)

##### perform bitwise NOT
print(~b)

##### perform bitwise left shift
print(a << 1)

##### perform bitwise right shift
print(a >> 1)
```

6. Membership Operators

```python
print(3 in [1,2,3,4])

print(5 in [1,3,4])

print(10 not in [1,2,3,4,5])

print(6 not in [1,3,5,6,7])
```

7.  Identity Operator

```python
a = 10
b = 10

print(a is b)

a = [1,2,3,4,5]
b = a
print(a is b)

a = [1,2,3]
b = [1,2,3]

print(a is b)
```

##### Operator Precedence

```python
result = 10 + 5 * 2
print(result)
```

```python
##### Using parenthesis
result = (10 + 5) * 2
print(result)
```

19. An ATM dispenses ₹500, ₹200 and ₹100 notes.

Write a Python program to calculate the minimum number of notes required to dispense a given amount, assuming the amount is a multiple of ₹100.

Use floor division and modulus operators.

```python
amount = 1800

notes_500 = amount // 500
remaining = amount % 500

notes_200 = remaining // 200
remaining = remaining % 200

notes_100 = remaining // 100
remaining = remaining % 100

print("₹500 Notes:", notes_500)
print("₹200 Notes:", notes_200)
print("₹100 Notes:", notes_100)

total_notes = notes_500 + notes_200 + notes_100

print("Total Notes:", total_notes)
```

20. A financial company identifies potentially risky transactions using the following conditions:
- Transaction amount exceeds ₹50,000.
- Transaction occurs outside normal business hours (before 9 AM or after 6 PM).
- Customer location differs from the registered location.
Flag the transaction if at least two of these conditions are satisfied.

Write a Python program using logical, comparison and arithmetic operators

```python
transaction_amount = 75000
transaction_hour = 20
customer_location = "Hyderabad"
registered_location = "Bangalore"

# Condition 1: Transaction amount exceeds ₹50,000
condition1 = transaction_amount > 50000

# Condition 2: Transaction occurs outside business hours
condition2 = transaction_hour < 9 or transaction_hour > 18

# Condition 3: Customer location differs from registered location
condition3 = customer_location != registered_location

# Count the number of satisfied conditions
risk_count = condition1 + condition2 + condition3

# Flag if at least two conditions are satisfied
if risk_count >= 2:
    print("High Risk Transaction")
else:
    print("Normal Transaction")
```
