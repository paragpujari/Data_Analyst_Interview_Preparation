##### Data Types

A Data types defines the type of value stored in the variable.

Python automatically determines the data type of the value once the value is assigned to it.

We can determine the type of the variable using type() function.

```python
a = 100
b = 10.5
c = 'Python'
d = (3+4j)
e = True

##### Determine the type of each variable
print(type(a))
print(type(b))
print(type(c))
print(type(d))
print(type(e))
```

##### Type Conversion

Type conversion is a technique used in Python that is used to convert the value from one data type to another.

##### Types

1.  Implicit Type Conversion

It refers to the process of automatically converting the value from one data type to another

```python
x = 10
y = 2.12
x = x+y
print(x)
```

##### OBSERVATIONS:

1.  Here x value was initially 10. Then the value of x got converted from 10 to 12.12 using Implicit Type Conversion.

2. Explicit Type Conversion

It refers to the process of manually converting the value from one data type to another using function


Function	                            Conversion
int()	    ------------>           Converts to integer
float()	    ------------>           Converts to float
str()	    ------------>           Converts to string
bool()	    ------------>           Converts to boolean
complex()	------------>           Converts to complex

```python
x = 10
ans = float(x)
print(ans)
```

##### Properties:

1.  Converting from lower data type to higher data type is feasible.(int ------> float)

2.  Converting from higher data type to lower data type leads to loss of precision.(float -----> int)

```python
##### 1. Converting from lower data type to higher data type is feasible.(int ------> float)

x = 10
print(x)

ans = float(x)
print(ans)
```

##### OBSERVATIONS:

1.  No loss of precision as value got converted from 10 ----> 10.0 (int -----> float)

```python
##### 2. Converting from higher data type to lower data type leads to loss of precision.(float -----> int)
x = 10.34
print(x)

ans = int(x)
print(ans)
```

##### OBSERVATIONS:

1.  Loss of precision takes place as 10.34 is changed to 10(.34 got lost after type conversion)

```python
##### int()	    ------------>           Converts to integer

x = 11.1
print(x)

ans = int(x)
print(ans)
```

```python
##### float()	    ------------>           Converts to float

x = 11
print(x)

ans = float(x)
print(ans)
```

```python
##### str()       ------------>        Converts to string
x = 10
print(x)

ans = str(x)
print(ans)
```

```python
##### 4. bool()        ------------>           Converts to boolean  
x = 1
print(x)

ans = bool(x)
print(ans)
```

```python
##### 5. Complex()      ------------->       Converts to complex
x = 1
print(x)

ans = complex(x)
print(ans)
```

20. A transaction processing system receives:

- Convert amount into float.
- Convert service charge into float.
- Convert transaction status into Boolean.
