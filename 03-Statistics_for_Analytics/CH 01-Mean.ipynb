##### Mean

1.  Mean is the average of all the observation values.

    It can also be defined as the sum of all the observations divided by number of all observations.




##### Solved Examples

1.  Calculate the mean of a dataset


```python
from statistics import mean

data = [10, 20, 30, 40, 50]

##### Calculate the mean of the above dataset

ans = mean(data)

print(f'Mean of the above dataset is {ans}')
```

2. Calculate the average employee salary

```python
salaries = [25000, 30000, 35000, 40000, 50000]

##### Calculate the mean of all the salaries

ans = sum(salaries)/len(salaries)

print(f'Mean of all the salaries is {ans}')
```

3. Calculate the average marks of students

```python
import numpy as np

marks = [80, 75, 90, 85, 70]

ans = np.mean(marks)

print(f'Mean of all the student marks is {ans}')
```

4. Calculate the mean of positive numbers only

```python
data = [10, -5, 20, -10, 30, 40]

filtered_data = [x for x in data if(x>0)]

print(filtered_data)

##### Calculate the mean of the filtered data

ans = sum(filtered_data)/len(filtered_data)

print(f'Mean of all the filtered data is {ans}')
```

5. Calculate the mean of two datasets combined

```python
data1 = [10, 20, 30]
data2 = [40, 50, 60]

data = data1 + data2

print(data)

from statistics import mean

ans = mean(data)

print(f'Mean of the combined two datasets is {ans}')
```

10. Calculate the mean of employee salaries department-wise

```python
employees = [
    {"name": "Amit", "department": "IT", "salary": 50000},
    {"name": "Rahul", "department": "IT", "salary": 60000},
    {"name": "Priya", "department": "HR", "salary": 40000},
    {"name": "Sneha", "department": "HR", "salary": 50000},
    {"name": "Raj", "department": "Finance", "salary": 70000},
    {"name": "Anita", "department": "Finance", "salary": 80000}
]

s_it=0
c_it=0

s_hr=0
c_hr=0

s_fi=0
c_fi=0

for x in employees:
    if(x['department']=='IT'):
        s_it += x['salary']
        c_it += 1
        avg_it = s_it/c_it
    elif(x['department']=='HR'):
        s_hr  += x['salary']
        c_hr  += 1
        avg_hr =s_hr/c_hr 
    else:
        s_fi += x['salary']
        c_fi += 1
        avg_fi = s_fi/c_fi 

print(f'Mean of salaries of all IT employees is {avg_it}')
print(f'Mean of salaries of all HR employees is {avg_hr}')
print(f'Mean of salaries of all Finance employees is {avg_fi}')
```

##### Analytical scenario-based interview question

1.  An organization has five employees with the following monthly salaries:

[25000, 30000, 35000, 40000, 45000]

The HR department increases the salary of one employee earning ₹25,000 to ₹75,000.

Answer the following:
- Calculate the original mean salary.
- Calculate the new mean salary after the increment.
- How much does the mean increase?
- Does increasing one employee's salary significantly affect the overall mean?
- Why can the mean be misleading when analyzing employee salaries if one employee earns an exceptionally high salary?

```python
from statistics import mean

data = [25000, 30000, 35000, 40000, 45000]

ans = mean(data)

print(f'Mean of all the employee salaries is {ans}')
```

```python
##### The HR department increases the salary of one employee earning ₹25,000 to ₹75,000


data = [75000, 30000, 35000, 40000, 45000]



ans1 = mean(data)

print(f'Mean of all the revised employee salary dataset is {ans1}')
```

##### OBSERVATIONS:

##### How much does the mean increase?

(a.)    After increasing the salary of one employee from 25000 to 75000, the mean of the employee salary has increased from 35000 to 45000.

##### Does increasing one employee's salary significantly affect the overall mean

- Increasing the salary of one employee has significantly increased the obverall mean.

- When the original salary was 25000 -----> overall mean of the dataset was 35000

- But when the salary of one employee increased from 25000 to 75000 -----> overall mean of the dataset got increased from 35000 to 45000.

##### Why can the mean be misleading when analyzing employee salaries if one employee earns an exceptionally high salary?

- This is because the mean is highly affected by outliers.
