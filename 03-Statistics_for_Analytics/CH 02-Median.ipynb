##### Median

Median refers to the process of finding out the middle values when all the elements are arranged in ascending or descending order.



##### Solved Examples

1.  Calculate the median of an odd-sized dataset

```python
from statistics import median

data = [10, 30, 20, 50, 40]

ans = median(data)

print(f'Median of the above dataset is {ans}')
```

2. Calculate the median of an even-sized dataset

```python
from statistics import median

data = [10,40,20,30]

ans = median(data)

print(f'Median of the above dataset is {ans}')
```

```python
data = [10, 20, 30, 40, 50, 60, 70, 80]

##### get the length of the dataset
N = len(data)

##### Calculate the median of the above dataset

##### Sort the data
data.sort()
print(data)

ans = (data[N//2-1] + data[N//2])//2


print(f'Median of the above dataset is {ans}')
```

##### Analytical Scenario Based Questions

1.  An organization has five employees with the following monthly salaries:

[25000, 30000, 35000, 40000, 45000]

The HR department increases the salary of one employee earning ₹25,000 to ₹1,00,000.

Answer the following:
- Calculate the original median salary.
- Calculate the new median salary after the increment.
- Does the median change after increasing one employee's salary?
- Why might the median be more reliable than the mean when analyzing employee salaries?
- If the organization has one highly paid executive earning ₹10,00,000, how would the median help HR understand the typical employee salary?

```python
##### Calculate the original median salary

salaries = [25000, 30000, 35000, 40000, 45000]

ans = median(salaries)

print(f'Median of the original employee salaries is {ans}')
```

```python
##### Calculate the new median salary after the increment. The HR department increases the salary of one employee earning ₹25,000 to ₹1,00,000

salaries_rev = [100000, 30000, 35000, 40000, 45000]

ans_rev = median(salaries_rev)

print(f'Median of the revised employee salaries is {ans_rev}')
```

##### Does the median change after increasing one employee's salary

- Yes, the median has increased by 5000(from 35000 to 40000) when the salary of one employee has been changed from 25000 to 100000

##### Why might the median be more reliable than the mean when analyzing employee salaries?

- Median be more reliable than the mean when analyzing employee salaries as it remains unaffected by the outliers.

##### If the organization has one highly paid executive earning ₹10,00,000, how would the median help HR understand the typical employee salary?

- The median helps HR understand the typical employee salary because it is less affected by exceptionally high or low salaries (outliers).
