8. Calculate the standard deviation of every numerical column in a Pandas DataFrame

```python
import pandas as pd

data = {
    "Age": [25, 30, 35, 40, 45],
    "Salary": [25000, 30000, 35000, 40000, 45000],
    "Experience": [1, 3, 5, 7, 9],
    "Performance_Score": [70, 75, 80, 85, 90]
}

df = pd.DataFrame(data)

std_age = stdev(df['Age'])
print(f'Standard deviation of the age column is {std_age:.2f}')

std_salary = stdev(df['Salary'])
print(f'Standard deviation of the salary column is {std_salary:.2f}')

std_experience = stdev(df['Experience'])
print(f'Standard deviation of the experience column is {std_experience:.2f}')

std_perfscore = stdev(df['Performance_Score'])
print(f'Standard deviation of the Performance Score column is {std_perfscore:.2f}')
```

9.  Calculate department-wise standard deviation of employee salaries

```python
import pandas as pd

data = {
    "Employee": ["A", "B", "C", "D", "E", "F", "G", "H", "I"],
    "Department": [
        "IT", "IT", "IT",
        "HR", "HR", "HR",
        "Finance", "Finance", "Finance"
    ],
    "Salary": [
        50000, 60000, 70000,
        30000, 35000, 40000,
        45000, 55000, 65000
    ]
}

df = pd.DataFrame(data)

print(df)

###### Calculate the department wise standard deviation salary
ans = df.groupby('Department')['Salary'].std(ddof=0)

print(ans)
```

10. Create a Python function that accepts a list and returns the following five statistical measures

```python
from statistics import mean, stdev, pstdev, pvariance, variance

def calculate_statistics(numbers):
    x = mean(numbers)
    y = pvariance(numbers)
    z = pstdev(numbers)
    a = variance(numbers)
    b = stdev(numbers)
    return x,y,z,a,b

numbers = [10, 20, 30, 40, 50, 60, 70]

x1,y1,z1,a1,b1 = calculate_statistics(numbers)

print(f'Mean value is {x1}')
print(f'Population variance is {y1}')
print(f'Population standard deviation is {z1}')
print(f'Sample variance is {a1}')
print(f'Sample standard deviation is {b1}')
```

##### Analytical Scenario Based Questions

1. A school wants to compare the consistency of students' examination results.


Answer the following:
- Calculate the mean marks of Class A.
- Calculate the mean marks of Class B.
- Calculate the population variance of both classes.
- Calculate the population standard deviation of both classes.
- Which class has more consistent performance?
- If both classes have the same average marks, can their standard deviations be different?
- Why is standard deviation useful when evaluating student performance?

```python
class_A = [70, 75, 80, 85, 90]
class_B = [40, 55, 70, 85, 100]

##### Calculate the mean marks of Class A
mean_A = mean(class_A)
print(f'Mean marks of Class A is {mean_A:.2f}')

##### Calculate the mean marks of Class B
mean_B = mean(class_B)
print(f'Mean marks of Class B is {mean_B:.2f}')

##### Calculate the population variance of both classes
popvar_A = pvariance(class_A)
print(f'Population Variance of Class A is {popvar_A:.2f}')

popvar_B = pvariance(class_B)
print(f'Population Variance of Class B is {popvar_B:.2f}')

##### Calculate the population standard deviation of both classes
popstd_A = pstdev(class_A)
print(f'Population Standard deviation of Class A is {popstd_A:.2f}')

popstd_B = pstdev(class_B)
print(f'Population Standard Deviation of Class B is {popstd_B:.2f}')

##### Which class has more consistent performance
if(popstd_A < popstd_B):
    print(f'Class A is having more consistent performance')
else:
    print(f'Class B is having more consistent performance')
```

##### If both classes have the same average marks, can their standard deviations be different

- Yes. Two classes can have the same average marks but different standard deviations because their marks may be distributed differently.

##### Why is standard deviation useful when evaluating student performance?

Standard deviation helps teachers understand:
- Consistency: Whether students are scoring similarly or their marks vary significantly.
- Performance variation: Whether marks are closely grouped around the average or widely spread.
- Identifying struggling students: A high spread may indicate students who need additional support.
- Comparing classes: Two classes may have the same average but different levels of consistency.
- Teaching effectiveness: It can help teachers identify whether learning outcomes are relatively uniform across students.
