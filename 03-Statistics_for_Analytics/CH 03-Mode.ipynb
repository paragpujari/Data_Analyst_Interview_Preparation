##### Mode

Mode is a technique of finding out the most frequent element frm the given dataset.

##### Types of Modes

- Unimodal ----->   Only one element can have the highest frequency
- Bimodal  ----->   Two elements can have the same highest frequency
- Multimodal --->   Two or more elements can have the same highest frequency
- No mode  ----->   Every value is having the same frequency

##### Solved Examples

1.  Calculate the mode using the statistics module

```python
from statistics import mode

data = [10, 20, 20, 30, 40, 20, 50]

ans = mode(data)

print(f'Mode of the above dataseries is {ans}')
```

2. Calculate the mode using Counter

```python
from collections import Counter

frequencies = Counter(data)

print(frequencies.most_common(1)[0][0])
```

3. Find all modes in a multimodal dataset

```python
data = [10, 10, 20, 20, 30, 40]

frequencies = Counter(data)

print(frequencies)

ans = frequencies.most_common(2)

print(ans[0][0], ans[1][0])
```

##### OBSERVATIONS:

(a).  The above dataset is a bimodal as same two values(10 and 20) are having the same highest frequencies.

4. Identify whether a dataset is unimodal, bimodal, or multimodal

```python
from collections import Counter 

data = [10, 10, 20, 20, 30, 30, 40]

frequencies = Counter(data)

max_frequency = max(frequencies.values())

c=0
for x,y in frequencies.items():
    if(y==max_frequency):
        c+=1
if(c==1):
    print(f'Unimodal')
elif(c==2):
    print(f'Bimodal')
else:
    print(f'Multimodal')
```

6. Calculate the mode of employee salaries and identify how many employees earn that salary

```python
from statistics import mode

employees = [
    {"name": "Amit", "salary": 40000},
    {"name": "Rahul", "salary": 50000},
    {"name": "Priya", "salary": 40000},
    {"name": "Sneha", "salary": 60000},
    {"name": "Raj", "salary": 50000},
    {"name": "Anita", "salary": 40000},
    {"name": "Vikram", "salary": 70000}
]

list=[]
for x in employees:
    list.append(x['salary'])
print(list)

##### Get the mode of all employee salaries
ans = mode(list)

print(f'Mode of all the employee salaries is {ans}')

lis1=[]
for x in employees:
    if(x['salary']==ans):
        lis1.append(x['name'])
print(lis1)
```

```python
data = [10, 20, 30, 20, 40, 20, 50,50,30, 30]

freq={}

for x in data:
    if(x in freq):
        freq[x] = freq[x] + 1
    else:
        freq[x] = 1
print(freq)


max_freq=0
for x,y in freq.items():
    if(y>max_freq):
        max_freq=y
        res=x        
print(f'Maximum occuring element is {res}')
```

10. Find all values having the maximum frequency in a dataset

```python
data = [10, 20, 20, 30, 30, 40, 40, 50]

freq={}

for x in data:
    if(x in freq):
        freq[x] = freq[x] + 1
    else:
        freq[x] = 1
print(freq)

max_freq = max(freq.values())
print(max_freq)

for x,y in freq.items():
    if(y==max_freq):
        print(x)
```

1.  An organization has seven employees with the following monthly salaries:

[30000, 40000, 30000, 50000, 40000, 30000, 60000]

The HR department increases the salary of one employee earning ₹30,000 to ₹1,00,000.

Answer the following:
- Calculate the original mode.
- Calculate the new mode after the salary increment.
- Does the mode change after the increment?
- How does the frequency of ₹30,000 change?
- Why might mode be more useful than mean when identifying the most common employee salary?

```python
salary = [30000, 40000, 30000, 50000, 40000, 30000, 60000]

##### original mode
print(f'Original mode is {mode(salary)}')
```

```python
##### new mode after the salary increment

salary = [100000, 40000, 30000, 50000, 40000, 30000, 60000]

##### Revised mode
print(f'Original mode is {mode(salary)}')
```

##### Does the mode change after the increment

- Yes the mode has changed after incrementing the salary of one employee from 30000 to 100000.

- It has increased from 30000 to 40000.

##### How does the frequency of ₹30,000 change?

- Initially the frequency of 30000 was 3. After updating the salary of 30000 of one employee from 30000 to 100000, the frequency has reduced from 3 to 2.

##### Why might mode be more useful than mean when identifying the most common employee salary?

- Mode be more useful than mean when identifying the most common employee salary as it remains unaffected by the extreme values.
