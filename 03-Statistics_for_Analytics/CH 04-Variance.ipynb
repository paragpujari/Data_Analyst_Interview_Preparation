9. Given employee salaries across three departments, calculate the variance department-wise

```python
data = {
    "Employee": ["A", "B", "C", "D", "E", "F", "G", "H", "I"],
    "Department": ["IT", "IT", "IT", "HR", "HR", "HR", "Finance", "Finance", "Finance"],
    "Salary": [50000, 60000, 70000, 30000, 35000, 40000, 45000, 55000, 65000]
}

df = pd.DataFrame(data)

print(df)

print('----------------------------------------------------------------------')

##### Calculate the variance of salaries department wise
ans = df.groupby('Department')['Salary'].var(ddof=0)

print(ans)
```

##### Analytical Scenario Based Questions

1.  An HR department has collected the following salaries from two departments.


Answer the following:
- Calculate the mean salary of Department A.
- Calculate the mean salary of Department B.
- Calculate the population variance of Department A.
- Calculate the population variance of Department B.
- Which department has more salary variation?
- Why might HR prefer a department with lower salary variance?
- Does lower variance necessarily mean better salary fairness? Explain.

```python
import pandas as pd

data = {
    "Employee": [1, 2, 3, 4, 5],
    "Department_A": [30000, 32000, 31000, 29000, 28000],
    "Department_B": [20000, 30000, 40000, 50000, 60000]
}

df = pd.DataFrame(data)

print(df)

##### Calculate the mean salary of Department A
mean_A = mean(df['Department_A'])
print(f'Mean salary of department A is {mean_A}')

##### Calculate the mean salary of Department B
mean_B = mean(df['Department_B'])
print(f'Mean salary of department B is {mean_B}')

##### Calculate the population variance of Department A
pvar_A = pvariance(df['Department_A'])
print(f'Population Variance of department A is {pvar_A}')

##### Calculate the population variance of Department B
pvar_B = pvariance(df['Department_B'])
print(f'Population Variance of department B is {pvar_B}')

##### Which department has more salary variation
if(pvar_A > pvar_B):
    print(f'Department A has more salary variation.')
else:
    print(f'Department B has more salary variation.')
```

##### Why might HR prefer a department with lower salary variance

HR might prefer a department with lower salary variance because:
- Salary Consistency: Employees receive relatively similar salaries.
- Employee Satisfaction: Reduces dissatisfaction caused by large salary differences.
- Pay Equity: May indicate more consistent compensation for employees in comparable roles.
- Budget Management: Makes salary planning and forecasting easier.
- Transparency: Helps HR maintain standardized salary structures.

##### Does lower variance necessarily mean better salary fairness? Explain

No, lower variance does not necessarily mean better salary fairness.

Lower variance indicates that employee salaries are closer to the average salary, but it does not guarantee that employees are paid fairly.

Example:
- Department A: Employees receive nearly identical salaries, but some employees have more experience and responsibilities.
- Department B: Employees receive different salaries based on their experience, skills, performance, and job responsibilities.
Department A has lower salary variance, but Department B might have a fairer salary structure.

HR should consider:
- Employee experience and qualifications.
- Job roles and responsibilities.
- Individual performance.
- Market salary standards.
- Equal pay for comparable work.
