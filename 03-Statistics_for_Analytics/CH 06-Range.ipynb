##### Range

1.  Range can be defined as the difference between the minimum and maximum value in the dataset.

##### Formula of Range:

Range = Maximum - Minimum

##### Solved Examples

1. Find out the range from the dataset [12, 15, 18, 22, 30]

```python
import numpy as np

data = [12, 15, 18, 22, 30]

max = np.max(data)
min = np.min(data)

ans = max-min

print(f'Range of the dataset is {ans}')
```

3.  Write a Python program to calculate the range without using max() and min().


```python
import numpy as np

numbers = [15, 25, 10, 45, 35]

max = numbers[0]

for x in numbers:
    if(x>max):
        max = x
print(f'Maximum element in the array is {max}')


min = numbers[0]

for x in numbers:
    if(x<min):
        min = x
print(f'Minimum element in the array is {min}')

##### Calculate the range of th dataset
ans = max-min
print(f'Range of the dataset is {ans}')
```

4. Write a Python program to calculate the range of each department's employee salaries

```python
departments = {
    "IT": [40000, 50000, 70000, 90000],
    "HR": [25000, 30000, 35000, 45000],
    "Finance": [35000, 45000, 60000, 80000]
}

for x,y in departments.items():
    print(x,':',np.max(y)-np.min(y))
```

5. Write a Python program to calculate the range of multiple datasets and identify which dataset has the highest range

```python
datasets = {
    "A": [10, 20, 30, 40],
    "B": [5, 15, 25, 100],
    "C": [20, 30, 40, 50]
}

for x,y in datasets.items():
    print(x,":",np.max(y)-np.min(y))
```

##### Analytical Scenario Based Questions

1.  A company has three departments.

Questions:
- Calculate the salary range for each department.
- Which department has the highest salary variation according to range?
- Which department has the lowest salary variation?
- Does Finance have an outlier?
- Why can range provide a misleading picture of salary distribution?

```python
departments = {
    "IT": [40000, 45000, 50000, 55000],
    "HR": [25000, 30000, 35000, 40000],
    "Finance": [30000, 40000, 60000, 120000]
}


##### Calculate the salary range for each department

for x,y in departments.items():
    print(x,":",np.max(y)-np.min(y))
```

##### Which department has the highest salary variation according to range

Ans  Finance department has the highst range.

##### Which department has the lowest salary variation

Ans Both IT and HR department has the same lowest range value.

##### Does Finance have an outlier

Ans Yes, Finance has an outlier as it creates a very large spread.

##### Why can range provide a misleading picture of salary distribution

Ans  Range can have a misleading salary distribution as it may consider outlier. It dos not care about how the salaries are distributed.
