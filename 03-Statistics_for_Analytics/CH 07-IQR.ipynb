##### InterQuartile Range

InterQuartile Rang(IQR) refers to the difference between third quartile(Q3) and first quartile(Q1).

##### Formula of IQR:

IQR = Q3 - Q1

##### Solved Examples

1.  Find out the IQR of the following dataset

```python
import numpy as np

data = [2, 4, 6, 8, 10, 12, 14, 16]

##### Calculate Q1
Q1 = np.quantile(data,0.25)
print(Q1)

##### Calculate Q3
Q3 = np.quantile(data,0.75)
print(Q3)

##### Caluclate IQR
IQR = Q3 - Q1
print(f'IQR is {IQR}')
```

3. Write a Python program to identify outliers using the IQR method

```python
import numpy as np

data = [10, 12, 14, 16, 18, 20, 22, 100]

##### Calculate Q1
Q1 = np.quantile(data,0.25)
print(Q1)

##### Calculate Q3
Q3 = np.quantile(data,0.75)
print(Q3)

##### Calculate IQR
IQR = Q3-Q1
print(f'IQR is {IQR}')

##### Calculate the lower and upper bound
lb = Q1-(1.5*IQR)
print(f'Lower bound is {lb}')

ub = Q3+(1.5*IQR)
print(f'Upper bound is {ub}')
```

##### OBSERVATIONS

(a.)    After getting the lower and the upper bound from the dataset, we came to the conclusion that 100 is the outlier.
