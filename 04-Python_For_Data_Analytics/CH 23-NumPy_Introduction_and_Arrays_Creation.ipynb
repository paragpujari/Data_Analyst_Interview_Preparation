#### Solved Examples

1.  Create a NumPy array from a Python list

```python
lists = [1,2,3,4,5]

##### Creation of numpy array
arr = np.array(lists)

print(arr)

print(f'Type of the array is {type(arr)}')
```

2. Create a 2D NumPy array

```python
numbers = [[1, 2, 3],
           [4, 5, 6]]

##### Creation of 2D numpy array
arr = np.array(numbers)

print(arr)

print(f'Type of the array is {type(arr)}')

print(f'Dimensions used in the array is {arr.ndim}')

print(f'Shape of the numpy array is {arr.shape}')
```

##### OBSERVATIONS:

(a.)  As the number of dimensions in the numpy array is 2, so it is a 2D numpy array.

3. Create an array of zeros and ones

```python
##### Creation of numpy arrays of zeros
ans_zeros = np.zeros(5)

print(ans_zeros)
print(f'No of dimensions in the array is {ans_zeros.ndim}')

##### Create a numpy array os ones
ans_ones = np.ones(5)

print(ans_ones)
print(f'No of dimensions in the array is {ans_ones.ndim}')
```

##### OBSERVATIONS:

(a.)  The above example creates a numpy array of zeros and ones which is one 1 dimension form.

##### Scenario: Employee salary data initialization

A company wants to initialize employee salary data for its HR analytics system.

You are given the following requirements:

```python
##### Display the shape, size, dimensions and data type of every relevant array

import numpy as np

# 1. Create an array containing salaries of 10 employees
salaries = np.array([30000, 35000, 40000, 45000, 50000,
                     55000, 60000, 65000, 70000, 75000])

print("1. Employee Salaries:")
print(salaries)


# 2. Initialize all 10 employee salaries with ₹30,000
initial_salaries = np.full(10, 30000)

print("\n2. Initialized Salaries:")
print(initial_salaries)


# 3. Generate 10 random performance scores between 1 and 100
performance = np.random.randint(1, 101, size=10)

print("\n3. Employee Performance Scores:")
print(performance)


# 4. Create a 2D array containing 5 employees and 3 columns
employee_data = np.array([
    [30000, 85, 101],
    [35000, 90, 102],
    [40000, 75, 103],
    [45000, 95, 104],
    [50000, 80, 105]
])

print("\n4. Employee Data (5 × 3):")
print(employee_data)


# 5. Generate 10 equally spaced salary values
salary_range = np.linspace(20000, 100000, 10)

print("\n5. Equally Spaced Salary Values:")
print(salary_range)


# 6. Display array properties
arrays = {
    "Salaries": salaries,
    "Initial Salaries": initial_salaries,
    "Performance Scores": performance,
    "Employee Data": employee_data,
    "Salary Range": salary_range
}

print("\n6. Array Properties:")

for name, arr in arrays.items():
    print(f"\n{name}")
    print("Shape:", arr.shape)
    print("Size:", arr.size)
    print("Dimensions:", arr.ndim)
    print("Data Type:", arr.dtype)
```
